```python
data["Price"]= data["Price"].str.replace("DH","")
data["Price"]= data["Price"].str.replace(" ","")
data["Price"]= data["Price"].str.replace("Prixàlabaisse","")

data.drop(index = data[data["Price"].isna()].index, inplace = True)

data.drop(index = data[data["Price"]=='Prixnonspécifié'].index, inplace = True)
data["Price"]= data["Price"].str.replace('Prixenhausse',"")
data.shape
```

```python
data = data.reindex()
data.index=range(data.shape[0])
data.index
```

```python
# Posting time
posting_time = list()
for i in range(len(data["Posting_time"])):
    pt = None
    date = str(data["Posting_time"][i])
    if date == "nan":
        pt = np.nan
        posting_time.append(pt)
            
    else : 
        month = int(date.split(",")[0].split("/")[0])
        #print(i, month)
        day = int(date.split(",")[0].split("/")[1])
        year = int(date.split(",")[0].split("/")[2])
        pt = (dt.datetime.now()-dt.datetime(year,month,day)).days
        posting_time.append(pt)#np.round(pt))
    
#max(posting_time)

data["posting_days"]=posting_time
data.head(10)
```

```python
# import data
df = pd.read_csv(r"C:\Users\hp\avito_cars_clean_data.csv")
df
```

```python
dummies_data = pd.get_dummies(df,prefix="",prefix_sep="" ,columns=['gearbox','power','origine'], dtype="int")
dummies_data
```

```python
#Apply label encoder on make and model 
le=LabelEncoder()
mark_model=df[["make","model"]]
#y=lambda x:le.fit_transform

encoded_df=mark_model.apply(le.fit_transform)
dummies_data[["make","model"]] = encoded_df
dummies_data
```

```python
dummies_data.rename(columns={"Automatique":"automatic",
                             "Manuelle":"manuel",
                             "Diesel":"diesel",
                             "Electrique":"electric",
                             "Essence" : "gasoline",
                             "Hybride" :"hybrid",
                             "Dédouanée": "cleared",
                             "Importée neuve":"imported_new",
                             "Pas encore dédouanée":"not_yet_cleared",
                             "WW au Maroc":"ww_in_morocco"},
                             inplace = True)
```

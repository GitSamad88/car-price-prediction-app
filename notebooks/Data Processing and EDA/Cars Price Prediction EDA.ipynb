## Mark And Model 

Q1: what are the top 20 Mark_model ?

```python
df[["mark_model"]].value_counts().to_frame().head(20).style.bar(color="red")
```

```python
def car_cloud(title_,col):
    text=""
    for word in col:
        text=text+" "+word
    wc=wordcloud.WordCloud(width=800,height=400)
    cloud=wc.generate(text)
    cloud.to_file("/downloads/car_cloud.png")
    return px.imshow(cloud,title = title_)
```

```python
Mark=[]
Model=[]
for word in df["mark_model"]:
    x=word.split()
    Mark.append(x[0])
    Model.append(" ".join(x[1:]))
```

```python
df["Mark"]=Mark
df["Model"]=Model
df.head()
```

```python
power_dummies=pd.get_dummies(df[["mark_model","Power"]],columns=["Power"])
```

```python
car_name,gas_and_diesel=[],[]
for m_m in power_dummies["mark_model"].unique():
    mm=power_dummies[power_dummies["mark_model"]==m_m]
    condition1 = mm["Power_Diesel"] == 1
    condition2 = mm["Power_Essence"] == 1
    if (condition1.any() or condition2.any()):
        car_name.append(m_m)
        gas_and_diesel.append("yes")
    else :
        car_name.append(m_m)
        gas_and_diesel.append("no")
        
        
```

```python
gas_and_diesel_car=pd.DataFrame({"Car":car_name,
                                 "is Gas and Diesel":gas_and_diesel})
gas_and_diesel_car[gas_and_diesel_car["is Gas and Diesel"]=="no"]
```
